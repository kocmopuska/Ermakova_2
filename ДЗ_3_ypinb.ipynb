## ***Самостоятельное задание № 1***

Постройте график функций:

*   Выберите две/три любую тригонометрические функции (например, math.sin(x)).
*   Создайте массив значений x и y, рассчитав значения функции для каждого x.
*   Используйте функцию plot из Matplotlib для построения графика зависимости y от x.
*   Добавьте заголовок и подписи к осям. Используйте plt.title, plt.xlabel, plt.ylabel.
*   Сохраните полученные графики в форматах (png, jpg, pdf).


````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2 * np.pi, 2 * np.pi, 400)

y_sin = np.sin(x)
y_cos = np.cos(x)
y_tanh = np.tanh(x)

plt.figure(figsize=(10, 6))

plt.plot(x, y_sin, label='sin(x)', color='blue', linewidth=2)
plt.plot(x, y_cos, label='cos(x)', color='orange', linewidth=2, linestyle='--')
plt.plot(x, y_tanh, label='tanh(x)', color='green', linewidth=2, linestyle='-.')

plt.title('Графики тригонометрических функций', fontsize=14, fontweight='bold')
plt.xlabel('Ось X (Радианы)', fontsize=12)
plt.ylabel('Ось Y (Значение функции)', fontsize=12)

plt.legend(fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)

plt.axhline(0, color='black', linewidth=0.8, linestyle='-')
plt.axvline(0, color='black', linewidth=0.8, linestyle='-')

plt.show()

plt.savefig('trig_plots.png', dpi=300, bbox_inches='tight')
plt.savefig('trig_plots.jpg', dpi=300, bbox_inches='tight')
plt.savefig('trig_plots.pdf', bbox_inches='tight')
````

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 3, 50)

y_squared = x**2
y_exp = np.exp(x)


plt.figure(figsize=(10, 6))


plt.plot(x, y_squared,
         label='y = x^2',
         linestyle='--',
         marker='o',
         color='red',
         linewidth=1.5,
         markersize=5)
plt.plot(x, y_exp,
         label='y = e^x',
         linestyle='-.',
         marker='^',
         color='green',
         linewidth=1.5,
         markersize=5)

plt.title('Графики функций $y = x^2$ и $y = e^x$', fontsize=14, fontweight='bold')
plt.xlabel('Ось X', fontsize=12)
plt.ylabel('Ось Y', fontsize=12)


plt.legend(fontsize=11)

plt.grid(True, linestyle=':', alpha=0.6)

plt.ylim(-1, 25)

plt.savefig('curves_plot.png', dpi=300, bbox_inches='tight')

plt.show()
````

````python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()

filename = list(uploaded.keys())[0]

# Simplified and more robust CSV reading
try:
    # Try reading with comma as separator and utf-8-sig encoding
    df = pd.read_csv(filename, sep=',', encoding='utf-8-sig')
except UnicodeDecodeError:
    # If UnicodeDecodeError occurs, try cp1251 encoding
    df = pd.read_csv(filename, sep=',', encoding='cp1251')
except Exception as e:
    # Fallback for any other unexpected errors, try without explicit encoding or sep inference
    print(f"Error reading CSV with explicit sep and encoding: {e}. Trying default parameters.")
    df = pd.read_csv(filename) # Let pandas infer sep and encoding

# Clean column names: strip whitespace and handle potential 'Unnamed' columns
df.columns = df.columns.str.strip()
df = df.loc[:, ~df.columns.str.contains('^Unnamed')] # Drop unnamed columns, often created from leading commas

epoch_col = [c for c in df.columns if 'epoch' in c.lower()][0]
train_loss_col = [c for c in df.columns if 'train' in c.lower() and 'loss' in c.lower()][0]
val_loss_col = [c for c in df.columns if 'val' in c.lower() and 'loss' in c.lower()][0]
map50_col = [c for c in df.columns if 'map50' in c.lower() and '95' not in c.lower()][0]
map95_col = [c for c in df.columns if 'map50-95' in c.lower() or 'map95' in c.lower()][0]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

ax1.plot(df[epoch_col], df[train_loss_col], color='red', linestyle='-', linewidth=2, label='Train Loss')
ax1.plot(df[epoch_col], df[val_loss_col], color='darkred', linestyle='--', linewidth=2, label='Val Loss')
ax1.set_title('Динамика функции потерь (Loss)', fontsize=14, fontweight='bold', pad=10)
ax1.set_xlabel('Эпоха (Epoch)', fontsize=12)
ax1.set_ylabel('Значение Loss', fontsize=12)
ax1.grid(True, linestyle=':', alpha=0.6)
ax1.legend(fontsize=11)

min_loss_idx = df[val_loss_col].idxmin()
min_loss_epoch = df.loc[min_loss_idx, epoch_col]
min_loss_val = df.loc[min_loss_idx, val_loss_col]

ax1.annotate(f'Min Loss: {min_loss_val:.4f}\n(Epoch {min_loss_epoch})',
             xy=(min_loss_epoch, min_loss_val),
             xytext=(min_loss_epoch + (df[epoch_col].max() * 0.1), min_loss_val + (df[val_loss_col].max() * 0.15)),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6),
             fontsize=10, fontweight='bold',
             bbox=dict(boxstyle='round,pad=0.3', fc='yellow', alpha=0.3))

ax2.plot(df[epoch_col], df[map50_col], color='green', linestyle='-', linewidth=2, label='mAP50')
ax2.plot(df[epoch_col], df[map95_col], color='darkgreen', linestyle='-.', linewidth=2, label='mAP50-95')
ax2.set_title('Метрики качества модели (mAP)', fontsize=14, fontweight='bold', pad=10)
ax2.set_xlabel('Эпоха (Epoch)', fontsize=12)
ax2.set_ylabel('Значение метрики', fontsize=12)
ax2.grid(True, linestyle=':', alpha=0.6)
ax2.legend(fontsize=11)

plt.suptitle('Показатели обученной модели из результатов', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
````

## ***Самостоятельное задание №5***

````python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()
filename = list(uploaded.keys())[0]

try:
    df = pd.read_csv(filename, sep=',', encoding='utf-8-sig')
except:
    df = pd.read_csv(filename, encoding='cp1251')

df.columns = df.columns.str.strip()

# Drop 'Unnamed: 0' column if it exists, as it's typically an artifact
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

# Identify the column containing data source names (e.g., 'Obelics', 'MINT-1T(HTML)')
# This column will provide the labels for the pie chart slices.
# Assuming 'DaraSource' is the column with these labels
data_source_label_col = 'DaraSource'

# Get the labels for the pie chart slices from the designated column
labels = df[data_source_label_col].tolist()

# Identify the columns that contain the numerical values for the pie charts.
# These are all columns except the data source label column.
models = [c for c in df.columns if c != data_source_label_col]

# Convert the 'models' columns to numeric, handling potential string issues (like ',' as decimal)
# This loop is crucial for ensuring the pie chart data is numerical.
for col_name in list(models): # Iterate over a copy to allow modification during iteration
    # Check if the column is of object type (string) and try to convert
    if df[col_name].dtype == 'object':
        try:
            df[col_name] = df[col_name].astype(str).str.replace(',', '.').astype(float)
        except ValueError:
            print(f"Warning: Column '{col_name}' contains non-numeric values that could not be converted to float. "
                  "This column will be skipped for plotting.")
            models.remove(col_name) # Remove column from models if it can't be converted
        except Exception as e:
            print(f"An unexpected error occurred while converting column '{col_name}': {e}. "
                  "This column will be skipped for plotting.")
            models.remove(col_name)

# Proceed with plotting only if there are models (numerical columns) to plot
if not models:
    print("No numerical columns found to plot pie charts after data cleaning.")
else:
    num_models = len(models)
    # Adjust figsize to prevent charts from being too small or overlapping
    # Using a max of 4 models per row to keep aspect ratio reasonable
    cols_per_row = 4
    rows = (num_models + cols_per_row - 1) // cols_per_row # Ceiling division
    fig, axes = plt.subplots(rows, min(num_models, cols_per_row), figsize=(7 * min(num_models, cols_per_row), 6 * rows))

    # Flatten axes array for easy iteration if there are multiple rows/cols
    if num_models == 1:
        axes = [axes] # Make it iterable if only one subplot
    else:
        axes = axes.flatten()

    cmap = plt.colormaps['Set3']
    colors = [cmap(i) for i in range(len(labels))]

    for i, model in enumerate(models):
        sizes = df[model].tolist()

        # Ensure sum of sizes is not zero to avoid error in pie chart if all values are 0
        if sum(sizes) == 0:
            axes[i].text(0.5, 0.5, 'No data to display', horizontalalignment='center', verticalalignment='center', transform=axes[i].transAxes, fontsize=12)
            axes[i].set_title(f'Распределение для\n{model}', fontsize=14, fontweight='bold', pad=15)
            axes[i].set_aspect('equal') # Keep aspect ratio square
            continue

        wedges, texts, autotexts = axes[i].pie(
            sizes,
            labels=labels, # Use the correct labels from 'DaraSource'
            autopct='%1.1f%%',
            startangle=140,
            colors=colors,
            textprops=dict(color="black")
        )

        plt.setp(autotexts, size=10, weight="bold")
        plt.setp(texts, size=11)
        axes[i].set_title(f'Распределение для\n{model}', fontsize=14, fontweight='bold', pad=15)

    # Hide unused subplots if num_models is not a perfect multiple of cols_per_row
    for j in range(len(models), len(axes)):
        fig.delaxes(axes[j])

    # Move legend to avoid overlapping with pie charts, especially with many subplots
    # Placing it outside the plot area
    fig.legend(wedges, labels, title="Категории / Токены", loc="center left", bbox_to_anchor=(0.95, 0.5), fontsize=12, title_fontsize=13)


    plt.suptitle('Сравнение состава моделей (Данные из mint-1T.csv)', fontsize=16, fontweight='bold', y=1.05)
    plt.tight_layout(rect=[0, 0, 0.9, 1]) # Adjust layout to make space for the legend
    plt.show()
````

Датасет: rcsi_science.csv

Построить гистограммы:

1) Построить гистограмму распределение количества вузов для Уральского федерального округа

2) Построить гистограмму распределения данных по региону с максимальным количеством вузов на Урале

3) Построить гистограмму распределения данных по отдельно взятому региону и году

Для каждой гистограммы создать отдельный контейнер (размер по данным), и форматирование (подписи осей, заголовок, цвет, подписи данных).

````python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()
filename = list(uploaded.keys())[0]

try:
    df = pd.read_csv(filename, sep=';', encoding='utf-8-sig')
except:
    try:
        df = pd.read_csv(filename, sep=',', encoding='utf-8-sig')
    except:
        df = pd.read_csv(filename, encoding='cp1251')

df.columns = df.columns.str.strip()

fd_col = 'federal_district' if 'federal_district' in df.columns else [c for c in df.columns if 'district' in c.lower() or 'округ' in c.lower()][0]
reg_col = 'region_name' if 'region_name' in df.columns else [c for c in df.columns if 'region' in c.lower() or 'регион' in c.lower()][0]
year_col = 'year' if 'year' in df.columns else [c for c in df.columns if 'year' in c.lower() or 'год' in c.lower()][0]

target_value_cols = [c for c in df.columns if c not in [fd_col, reg_col, year_col] and pd.api.types.is_numeric_dtype(df[c])]
val_col = target_value_cols[0] if target_value_cols else df.columns[-1]

df_urfo = df[df[fd_col].astype(str).str.contains('ural|урал', case=False, na=False)].copy()

if not df_urfo.empty:
    fig1, ax1 = plt.subplots(figsize=(10, 6))
    counts1, bins1, patches1 = ax1.hist(df_urfo[val_col], bins=8, color='skyblue', edgecolor='black', alpha=0.8)
    ax1.set_title(f'Распределение показателя ({val_col}) в Уральском ФО', fontsize=14, fontweight='bold', pad=15)
    ax1.set_xlabel(f'Значение ({val_col})', fontsize=12)
    ax1.set_ylabel('Частота (Количество наблюдений)', fontsize=12)
    ax1.grid(True, linestyle='--', alpha=0.5)
    ax1.bar_label(patches1, fmt='%.0f', padding=3, fontsize=10, fontweight='bold')
    plt.tight_layout()
    plt.show()

    max_region = df_urfo.groupby(reg_col)[val_col].max().idxmax()
    df_max_region = df_urfo[df_urfo[reg_col] == max_region].copy()

    fig2, ax2 = plt.subplots(figsize=(10, 6))
    counts2, bins2, patches2 = ax2.hist(df_max_region[val_col], bins=6, color='salmon', edgecolor='black', alpha=0.8)
    ax2.set_title(f'Распределение по региону с макс. показателем ({max_region})', fontsize=14, fontweight='bold', pad=15)
    ax2.set_xlabel(f'Значение ({val_col})', fontsize=12)
    ax2.set_ylabel('Частота', fontsize=12)
    ax2.grid(True, linestyle='--', alpha=0.5)
    ax2.bar_label(patches2, fmt='%.0f', padding=3, fontsize=10, fontweight='bold')
    plt.tight_layout()
    plt.show()

    chosen_region = df_urfo[reg_col].unique()[0]
    chosen_year = df_urfo[year_col].unique()[0]
    df_target = df_urfo[(df_urfo[reg_col] == chosen_region) & (df_urfo[year_col] == chosen_year)].copy()

    fig3, ax3 = plt.subplots(figsize=(8, 5))
    counts3, bins3, patches3 = ax3.hist(df_target[val_col], bins=4, color='mediumaquamarine', edgecolor='black', alpha=0.8)
    ax3.set_title(f'Распределение: {chosen_region} ({chosen_year} год)', fontsize=14, fontweight='bold', pad=15)
    ax3.set_xlabel(f'Значение ({val_col})', fontsize=12)
    ax3.set_ylabel('Частота', fontsize=12)
    ax3.grid(True, linestyle='--', alpha=0.5)
    ax3.bar_label(patches3, fmt='%.0f', padding=3, fontsize=10, fontweight='bold')
    plt.tight_layout()
    plt.show()
else:
    print("Данные по Уральскому федеральному округу не найдены. Проверьте содержимое файла.")
````
