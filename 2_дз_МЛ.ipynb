### **Самостоятельная работа №3:**

* Загрузите данные из файла movies_small_df
* Сохраните в отдельные переменные текстовый и числовые столбцы
* Выведите объем занимаемой памяти для текстовых и числовых столбцов
* Установите тип 'category' для какого-либо столбца
* Сохраните его в отдельную переменную
* Сравните столбец до и после преобразования

```python
# Ваш код
import pandas as pd
from google.colab import files

uploaded = files.upload()
my_movies = pd.read_excel('movies_small_df.xlsx')

text_columns = my_movies.select_dtypes(include=['object'])

numeric_columns = my_movies.select_dtypes(include=['number'])

text_memory = text_columns.memory_usage(deep=True).sum()
numeric_memory = numeric_columns.memory_usage(deep=True).sum()

print(f"Текстовые столбцы занимают: {text_memory} байт")
print(f"Числовые столбцы занимают: {numeric_memory} байт\n")


genre_mem_before = my_movies['Genre'].memory_usage(deep=True)

genre_before = my_movies['Genre'].copy()

my_movies['Genre'] = my_movies['Genre'].astype('category')

genre_category = my_movies['Genre']

genre_mem_after = genre_category.memory_usage(deep=True)

print(" Сравнение столбца 'Genre'")
print(f"Тип данных ДО: {genre_before.dtype}")
print(f"Тип данных ПОСЛЕ: {genre_category.dtype}")
print(f"Память ДО преобразования: {genre_mem_before} байт")
print(f"Память ПОСЛЕ преобразования: {genre_mem_after} байт")
print(f"Оптимизация памяти: {round((1 - genre_mem_after/genre_mem_before)*100, 2)}%")
```
